```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)

master_df = pd.read_csv(
    "../data/processed/master_dataset.csv"
)

master_df['DATE_TIME'] = pd.to_datetime(
    master_df['DATE_TIME']
)

print("Master dataset loaded successfully")

display(master_df.head())
```

```python
master_df['HOUR'] = master_df['DATE_TIME'].dt.hour

master_df['DAY'] = master_df['DATE_TIME'].dt.day

master_df['MONTH'] = master_df['DATE_TIME'].dt.month

master_df['DAY_OF_WEEK'] = master_df['DATE_TIME'].dt.dayofweek

master_df['WEEK_OF_YEAR'] = master_df['DATE_TIME'].dt.isocalendar().week

print("Time features created successfully")

display(master_df.head())
```

```python
master_df['IS_PEAK_HOUR'] = np.where(
    (master_df['HOUR'] >= 10) &
    (master_df['HOUR'] <= 15),
    1,
    0
)

display(master_df.head())
```

```python
master_df['TEMP_DIFFERENCE'] = (
    master_df['MODULE_TEMPERATURE']
    -
    master_df['AMBIENT_TEMPERATURE']
)

display(master_df.head())
```

```python
master_df['IRRADIATION_EFFICIENCY'] = (
    master_df['AC_POWER']
    /
    (master_df['IRRADIATION'] + 0.0001)
)

display(master_df.head())
```

```python
master_df = master_df.sort_values(
    by='DATE_TIME'
)

master_df['PREVIOUS_AC_POWER'] = (
    master_df['AC_POWER'].shift(1)
)

display(master_df.head())
```

```python
master_df.dropna(inplace=True)

print(master_df.isnull().sum())
```

```python
master_df['ROLLING_MEAN_AC_POWER'] = (
    master_df['AC_POWER']
    .rolling(window=3)
    .mean()
)

display(master_df.head())
```

```python
master_df.dropna(inplace=True)

print(master_df.isnull().sum())
```

```python
print(master_df.columns)

print("\nFinal Shape:")
print(master_df.shape)
```

```python
master_df.to_csv(
    "../data/processed/featured_dataset.csv",
    index=False
)

print("Featured dataset saved successfully")
```
