```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

print("Libraries Loaded Successfully")
```

```python
gen1 = pd.read_csv("../data/raw/plant1_generation.csv")
weather1 = pd.read_csv("../data/raw/plant1_weather.csv")

gen2 = pd.read_csv("../data/raw/plant2_generation.csv")
weather2 = pd.read_csv("../data/raw/plant2_weather.csv")

print("All datasets loaded successfully")
```

```python
print("Plant 1 Generation")
display(gen1.head())

print("Plant 1 Weather")
display(weather1.head())

print("Plant 2 Generation")
display(gen2.head())

print("Plant 2 Weather")
display(weather2.head())
```

```python
print("Plant 1 Generation Info")
gen1.info()

print("\nPlant 1 Weather Info")
weather1.info()

print("\nPlant 2 Generation Info")
gen2.info()

print("\nPlant 2 Weather Info")
weather2.info()
```

```python
print("Plant 1 Generation Missing Values")
print(gen1.isnull().sum())

print("\nPlant 1 Weather Missing Values")
print(weather1.isnull().sum())

print("\nPlant 2 Generation Missing Values")
print(gen2.isnull().sum())

print("\nPlant 2 Weather Missing Values")
print(weather2.isnull().sum())
```

```python
print(gen1.dtypes)

print("\n")
print(weather1.dtypes)
```

```python
gen1['DATE_TIME'] = pd.to_datetime(gen1['DATE_TIME'])
weather1['DATE_TIME'] = pd.to_datetime(weather1['DATE_TIME'])

gen2['DATE_TIME'] = pd.to_datetime(gen2['DATE_TIME'])
weather2['DATE_TIME'] = pd.to_datetime(weather2['DATE_TIME'])

print("DATE_TIME converted successfully")
```

```python
print(gen1.dtypes)

print("\n")
print(weather1.dtypes)
```

```python
plant1 = pd.merge(
    gen1,
    weather1,
    on=['DATE_TIME', 'PLANT_ID'],
    how='inner'
)

print("Plant 1 merged successfully")

print("Shape:", plant1.shape)

display(plant1.head())
```

```python
plant1.drop(columns=['SOURCE_KEY_y'], inplace=True)

plant1.rename(
    columns={'SOURCE_KEY_x': 'SOURCE_KEY'},
    inplace=True
)

display(plant1.head())
```

```python
plant2 = pd.merge(
    gen2,
    weather2,
    on=['DATE_TIME', 'PLANT_ID'],
    how='inner'
)

plant2.drop(columns=['SOURCE_KEY_y'], inplace=True)

plant2.rename(
    columns={'SOURCE_KEY_x': 'SOURCE_KEY'},
    inplace=True
)

print("Plant 2 merged successfully")

print("Shape:", plant2.shape)

display(plant2.head())
```

```python
master_df = pd.concat(
    [plant1, plant2],
    ignore_index=True
)

print("Master Dataset Created")

print("Master Shape:", master_df.shape)

display(master_df.head())
```

```python
print(master_df.columns)
```

```python
master_df.to_csv(
    "../data/processed/master_dataset.csv",
    index=False
)

print("Master dataset saved successfully")
```
