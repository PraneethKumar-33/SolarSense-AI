```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

master_df = pd.read_csv(
    "../data/processed/featured_dataset.csv"
)

master_df['DATE_TIME'] = pd.to_datetime(
    master_df['DATE_TIME']
)

print("Featured dataset loaded successfully")

display(master_df.head())
```

```python
print("Dataset Shape:")
print(master_df.shape)

print("\nColumns:")
print(master_df.columns)

print("\nStatistical Summary:")
display(master_df.describe())
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    master_df['AC_POWER'],
    bins=50,
    kde=True
)

plt.title("Distribution of AC Power")

plt.xlabel("AC Power")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(16,6))

plt.plot(
    master_df['DATE_TIME'],
    master_df['AC_POWER']
)

plt.title("AC Power Over Time")

plt.xlabel("Date Time")

plt.ylabel("AC Power")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=master_df['IRRADIATION'],
    y=master_df['AC_POWER'],
    alpha=0.5
)

plt.title("Irradiation vs AC Power")

plt.xlabel("Irradiation")

plt.ylabel("AC Power")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=master_df['MODULE_TEMPERATURE'],
    y=master_df['AC_POWER'],
    alpha=0.5
)

plt.title("Module Temperature vs AC Power")

plt.xlabel("Module Temperature")

plt.ylabel("AC Power")

plt.show()
```

```python
hourly_power = master_df.groupby(
    'HOUR'
)['AC_POWER'].mean()

plt.figure(figsize=(12,6))

sns.lineplot(
    x=hourly_power.index,
    y=hourly_power.values,
    marker='o'
)

plt.title("Average AC Power by Hour")

plt.xlabel("Hour")

plt.ylabel("Average AC Power")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(14,10))

correlation_matrix = master_df.corr(
    numeric_only=True
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
correlation_with_target = correlation_matrix[
    'AC_POWER'
].sort_values(
    ascending=False
)

print(correlation_with_target)
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    x=master_df['AC_POWER']
)

plt.title("AC Power Outlier Detection")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    x=hourly_power.index,
    y=hourly_power.values,
    marker='o'
)

plt.title("Average AC Power by Hour")

plt.savefig(
    "../outputs/graphs/hourly_ac_power.png"
)

plt.show()
```
